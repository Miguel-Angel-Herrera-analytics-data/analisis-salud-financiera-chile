```python
import pandas as pd

# Diccionarios compartidos
columnas_clave = {
    'Efectivoyequivalentesalefectivo': 'Efectivo',
    'Activoscorrientestotales': 'Activos_Corrientes',
    'Totaldeactivosnocorrientes': 'Activos_No_Corrientes',
    'Pasivoscorrientestotales': 'Pasivos_Corrientes',
    'Totaldepasivos': 'Pasivos_Totales',
    'Patrimoniototal': 'Patrimonio_Total',
    'Ingresosdeactividadesordinarias': 'Ingresos',
    'Gananciabruta': 'Ganancia_Bruta',
    'Ganancia(pérdida)': 'Utilidad_Neta'
}

cols_numericas = ['Efectivo', 'Activos_Corrientes', 'Activos_No_Corrientes',
                  'Pasivos_Corrientes', 'Pasivos_Totales', 'Patrimonio_Total',
                  'Ingresos', 'Ganancia_Bruta', 'Utilidad_Neta']

ratios = ['Empresa', 'Año', 'Liquidez_Corriente', 'Deuda_Patrimonio',
          'ROE', 'Margen_Bruto', 'Margen_Neto']

def procesar_empresa(archivo, nombre, año=2024):
    df_raw = pd.read_excel(archivo, header=12)
    df = df_raw[list(columnas_clave.keys())].copy()
    df.rename(columns=columnas_clave, inplace=True)
    df['Empresa'] = nombre
    df['Año'] = año
    for col in cols_numericas:
        df[col] = pd.to_numeric(df[col], errors='coerce')
    df['Activos_Totales'] = df['Activos_Corrientes'] + df['Activos_No_Corrientes']
    df['Liquidez_Corriente'] = (df['Activos_Corrientes'] / df['Pasivos_Corrientes']).round(2)
    df['Deuda_Patrimonio'] = (df['Pasivos_Totales'] / df['Patrimonio_Total']).round(2)
    df['ROE'] = (df['Utilidad_Neta'] / df['Patrimonio_Total'] * 100).round(2)
    df['Margen_Bruto'] = (df['Ganancia_Bruta'] / df['Ingresos'] * 100).round(2)
    df['Margen_Neto'] = (df['Utilidad_Neta'] / df['Ingresos'] * 100).round(2)
    return df

# Procesar las tres empresas
df_falabella = procesar_empresa('falabella_financiero.xlsx', 'Falabella')
df_cencosud  = procesar_empresa('cencosud_financiero.xlsx', 'Cencosud')
df_ripley    = procesar_empresa('ripley_financiero.xlsx', 'Ripley')

# Unir todo
df_retail = pd.concat([df_falabella[ratios],
                       df_cencosud[ratios],
                       df_ripley[ratios]], ignore_index=True)

print(df_retail.to_string(index=False))
```

```python
import matplotlib.pyplot as plt

# Configuración general
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
fig.suptitle('Análisis de Salud Financiera - Retail Chile 2024',
             fontsize=16, fontweight='bold')

empresas = df_retail['Empresa']
colores = ['#2196F3', '#4CAF50', '#FF5722']

# 1. Liquidez Corriente
axes[0,0].bar(empresas, df_retail['Liquidez_Corriente'], color=colores)
axes[0,0].axhline(y=1, color='red', linestyle='--', label='Mínimo aceptable')
axes[0,0].set_title('Liquidez Corriente')
axes[0,0].set_ylabel('Veces')
axes[0,0].legend()

# 2. Deuda/Patrimonio
axes[0,1].bar(empresas, df_retail['Deuda_Patrimonio'], color=colores)
axes[0,1].set_title('Deuda / Patrimonio')
axes[0,1].set_ylabel('Veces')

# 3. ROE
axes[0,2].bar(empresas, df_retail['ROE'], color=colores)
axes[0,2].set_title('ROE (%)')
axes[0,2].set_ylabel('%')

# 4. Margen Bruto
axes[1,0].bar(empresas, df_retail['Margen_Bruto'], color=colores)
axes[1,0].set_title('Margen Bruto (%)')
axes[1,0].set_ylabel('%')

# 5. Margen Neto
axes[1,1].bar(empresas, df_retail['Margen_Neto'], color=colores)
axes[1,1].set_title('Margen Neto (%)')
axes[1,1].set_ylabel('%')

# 6. Tabla resumen
axes[1,2].axis('off')
tabla = axes[1,2].table(
    cellText=df_retail[['Empresa','Liquidez_Corriente','Deuda_Patrimonio',
                         'ROE','Margen_Bruto','Margen_Neto']].values,
    colLabels=['Empresa','Liquidez','D/P','ROE%','M.Bruto%','M.Neto%'],
    cellLoc='center',
    loc='center'
)
tabla.auto_set_font_size(False)
tabla.set_fontsize(9)

plt.tight_layout()
plt.savefig('analisis_retail_chile_2024.png', dpi=150, bbox_inches='tight')
plt.show()
print("Gráfico guardado!")
```

```python
# Procesar todos los archivos por empresa y año
archivos = [
    ('falabella_2022.xlsx', 'Falabella', 2022),
    ('falabella_2023.xlsx', 'Falabella', 2023),
    ('falabella_financiero.xlsx', 'Falabella', 2024),
    ('falabella_2025.xlsx', 'Falabella', 2025),
    ('cencosud_2022.xlsx', 'Cencosud', 2022),
    ('cencosud_2023.xlsx', 'Cencosud', 2023),
    ('cencosud_financiero.xlsx', 'Cencosud', 2024),
    ('cencosud_2025.xlsx', 'Cencosud', 2025),
    ('ripley_2022.xlsx', 'Ripley', 2022),
    ('ripley_2023.xlsx', 'Ripley', 2023),
    ('ripley_financiero.xlsx', 'Ripley', 2024),
    ('ripley_2025.xlsx', 'Ripley', 2025),
]

# Procesar y unir todo
dfs = []
for archivo, empresa, año in archivos:
    df_temp = procesar_empresa(archivo, empresa, año)
    dfs.append(df_temp[ratios])
    print(f"✓ {empresa} {año} procesado")

# DataFrame final con todos los años y empresas
df_retail_completo = pd.concat(dfs, ignore_index=True)

print("\n=== DATASET COMPLETO ===")
print(df_retail_completo.to_string(index=False))
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 3, figsize=(16, 9))
fig.suptitle('Evolución Salud Financiera - Retail Chile 2022-2025',
             fontsize=16, fontweight='bold')

empresas = ['Falabella', 'Cencosud', 'Ripley']
colores = {'Falabella': '#2196F3', 'Cencosud': '#4CAF50', 'Ripley': '#FF5722'}
ratios_graficos = ['Liquidez_Corriente', 'Deuda_Patrimonio', 'ROE',
                   'Margen_Bruto', 'Margen_Neto']
titulos = ['Liquidez Corriente', 'Deuda / Patrimonio', 'ROE (%)',
           'Margen Bruto (%)', 'Margen Neto (%)']

for idx, (ratio, titulo) in enumerate(zip(ratios_graficos, titulos)):
    ax = axes[idx//3, idx%3]
    for empresa in empresas:
        df_emp = df_retail_completo[df_retail_completo['Empresa'] == empresa]
        ax.plot(df_emp['Año'], df_emp[ratio],
                marker='o', label=empresa,
                color=colores[empresa], linewidth=2)
    ax.set_title(titulo, fontweight='bold')
    ax.set_xlabel('Año')
    ax.legend()
    ax.grid(True, alpha=0.3)

# Celda vacía
axes[1,2].axis('off')

plt.tight_layout()
plt.savefig('tendencias_retail_chile_2022_2025.png', dpi=150, bbox_inches='tight')
plt.show()
print("Gráfico de tendencias guardado!")
```
